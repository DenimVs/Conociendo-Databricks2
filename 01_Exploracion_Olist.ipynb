```python
# =============================================================================
# MÓDULO 1 - INGESTA DE DATOS
# Notebook: 01_Exploracion_Olist
#
# Objetivo:
# Leer el archivo de clientes desde un Volume de Databricks utilizando PySpark.
# =============================================================================


# -----------------------------------------------------------------------------
# Lectura del archivo CSV
# -----------------------------------------------------------------------------
# spark.read      -> Utiliza el motor de procesamiento Apache Spark.
# csv()           -> Indica que el origen de los datos es un archivo CSV.
# header=True     -> La primera fila contiene los nombres de las columnas.
# inferSchema=True-> Spark intenta detectar automáticamente el tipo de dato
#                    de cada columna (String, Integer, Date, etc.).
# -----------------------------------------------------------------------------

customers_df = spark.read.csv(
    "/Volumes/workspace/default/raw/olist_customers_dataset.csv",
    header=True,
    inferSchema=True
)


# -----------------------------------------------------------------------------
# Visualizar el DataFrame
# -----------------------------------------------------------------------------
# display() es una función propia de Databricks.
# Muestra los datos en formato de tabla interactiva.
# Es equivalente a un "vista previa" del DataFrame.
# -----------------------------------------------------------------------------

display(customers_df)


# -----------------------------------------------------------------------------
# Mostrar el esquema del DataFrame
# -----------------------------------------------------------------------------
# printSchema() permite conocer:
#
# • Nombre de cada columna.
# • Tipo de dato detectado por Spark.
# • Si la columna acepta valores nulos (nullable).
#
# Es una de las primeras verificaciones que siempre se realizan
# después de cargar un dataset.
# -----------------------------------------------------------------------------

customers_df.printSchema()
```

```python
# =============================================================================
# INSPECCIÓN INICIAL DEL DATASET
# =============================================================================

# -----------------------------------------------------------------------------
# Cantidad de filas y columnas
# -----------------------------------------------------------------------------

print(f"Filas: {customers_df.count()}")
print(f"Columnas: {len(customers_df.columns)}")


# -----------------------------------------------------------------------------
# Nombre de las columnas
# -----------------------------------------------------------------------------

print(customers_df.columns)
```

```python
# =============================================================================
# VALIDACIÓN DE IDENTIFICADORES
# =============================================================================

# Cantidad total de registros
print(f"Total registros: {customers_df.count()}")

# Cantidad de customer_id distintos
print(f"customer_id únicos: {customers_df.select('customer_id').distinct().count()}")

# Cantidad de customer_unique_id distintos
print(f"customer_unique_id únicos: {customers_df.select('customer_unique_id').distinct().count()}")
```
